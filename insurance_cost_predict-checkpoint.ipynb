```python
# Import the files
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Import the insurance DB and dividing the data in dependent (X) and independent(y)
dataset = pd.read_csv("insurance.csv")
X = dataset.iloc[:,0:-1].values
y = dataset.iloc[:,-1].values
```

```python
X[0:15]
```

```python
y
```

```python
dataset.info()
```

```python
y[5:6]
```

```python
#Applying Categorical Encoding on Male/Female and Region(the last column)
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers= [('encoder',OneHotEncoder(drop='first', sparse_output=False),[1,-1])], remainder = 'passthrough')
X = np.array(ct.fit_transform(X))
```

```python
X[0:15,:]
```

```python
#Applying label Encoder on the column "smoker" (yes/no)
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X[:,7] = np.array(le.fit_transform(X[:,7]))
```

```python
X[0:5]
```

```python
# Splitting the data in Test Set and Training set
from sklearn.model_selection import train_test_split
X_train,X_test, y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state = 0)
```

```python
#Applying Feature Scaling to the data only on the column BMI, number of Children and Age
#Rest of the columns are Categorically encoded or applied LabelEncoder
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train[:,4:7] = sc.fit_transform(X_train[:,4:7])
X_test[:,4:7] = sc.transform(X_test[:,4:7])
```

```python
X_train[1:10,:]
```

```python
y_train
```

```python
# Beggining of testing with various models to see which one fits the best.
```

```python
#Applying Multiple Linear regression on the training/test set
from sklearn.linear_model import LinearRegression
multi_reg = LinearRegression()
multi_reg.fit(X_train,y_train)
```

```python
y_pred = multi_reg.predict(X_test)
# np.set_printoptions(precision=2)
# print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
from sklearn.metrics import r2_score
r2_score(y_test, y_pred)
```

```python
plt.scatter(X_test[:,5],y_test,color ='red')
plt.scatter(X_test[:,5],y_pred,color = 'blue')
plt.title('Insurance Information')
plt.xlabel('BMI')
plt.ylabel('Insurance')
plt.show()
```

```python
plt.scatter(X_test[:,4],y_test,color ='red')
plt.scatter(X_test[:,4],y_pred,color = 'blue')
plt.title('Insurance Information')
plt.xlabel('Age')
plt.ylabel('Insurance')
plt.show()
```

```python
# Conclusion on multiple linear regression : Accurace of 79.99%
```

```python
# Implementing Support Vector Regression SVR
# For SVR, we need to scale X and Y if there is huge difference in the value of X and Y
#X is already scaled above
# Reshaping y_train to 2d, since standard scalar accepts 2D array
y_train_2d = y_train.reshape(len(y_train),1)
```

```python
#Also reshaping y_test to 2D for standard Scalar
y_test_2d = y_test.reshape(len(y_test),1)
```

```python
#Applying Feature Scaling to Dependent Variable
sc_y = StandardScaler()
sc_y_train = sc_y.fit_transform(y_train_2d)
sc_y_test = sc_y.transform(y_test_2d)
#arr_1d = arr_2d.reshape(-1)
sc_y_train = sc_y_train.reshape(-1)
sc_y_test = sc_y_test.reshape(-1)
```

```python
#Training the SVR model on the training dataset
from sklearn.svm import SVR
regressor_svr = SVR(kernel = 'rbf')
regressor_svr.fit(X_train,sc_y_train)
```

```python
#Predicting Test Result
y_pred = sc_y.inverse_transform(regressor_svr.predict(X_test).reshape(-1,1))
```

```python
y_pred[0:10]
```

```python
#Checking accuracy score in SVR
from sklearn.metrics import r2_score
r2_score(y_test, y_pred)
```

```python
plt.scatter(X_test[:,5],y_test,color ='red')
plt.scatter(X_test[:,5],y_pred,color = 'blue')
plt.title('Insurance Information for SVR')
plt.xlabel('BMI')
plt.ylabel('Insurance')
plt.show()
## Result is Similar to Polynomial Regression, Polynomial Regression is 78%, SVR is 77% accuracy
```

```python
plt.scatter(X_test[:,4],y_test,color ='red')
plt.scatter(X_test[:,4],y_pred,color = 'blue')
plt.title('Insurance Information')
plt.xlabel('Age')
plt.ylabel('Insurance')
plt.show()
```

```python
#Conclusion for SVR, accuracy score is 88.8%, which is best so far.
```
